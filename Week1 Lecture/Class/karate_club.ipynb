# Zachary's Karate Club

Load the built-in karate club graph, visualize it, and summarize basic statistics.

```python
import networkx as nx
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

G = nx.karate_club_graph()

print(f"Number of nodes: {G.number_of_nodes()}")
print(f"Number of edges: {G.number_of_edges()}")

degrees = dict(G.degree())
top_five = sorted(degrees.items(), key=lambda item: item[1], reverse=True)[:5]
print("\nFive highest-degree nodes:")
for node, degree in top_five:
    print(f"  Node {node}: degree {degree}")
```

```python
plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G, seed=42)
nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="lightblue",
    node_size=400,
    font_size=8,
    edge_color="gray",
)
plt.title("Zachary's Karate Club")
plt.axis("off")
plt.tight_layout()
plt.show()
```

## Hand verification

Pick one node and compare its reported degree to a manual count from the adjacency structure.

```python
verify_node = 0
reported_degree = G.degree(verify_node)
neighbors = list(G.neighbors(verify_node))
manual_count = len(neighbors)

print(f"Node {verify_node}")
print(f"  Reported degree: {reported_degree}")
print(f"  Neighbors: {neighbors}")
print(f"  Manual count (len(neighbors)): {manual_count}")
print(f"  Match: {reported_degree == manual_count}")
```
